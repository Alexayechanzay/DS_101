```python
import math
import numpy as np
print(np.__version__)
```

```python
# correlation coefficient (r)
def correlation_coefficient(n, sum_of_xy, sum_of_x, sum_of_y, sum_of_xs, sum_of_ys):
    numerator = (n * sum_of_xy) - (sum_of_x * sum_of_y)
    deno1 = ((n * sum_of_xs)-(sum_of_x ** 2)) * ((n * sum_of_ys)-(sum_of_y ** 2))
    deno2 = math.sqrt(deno1)
    r = numerator / deno2

    if r > 0.97:
        print("Correlation coefficient is near 1\nr: ",r)
    else:
        print(r)
```

```python
# y-intercept (b)
def b(sum_of_y, m_slope, sum_of_x,n):
    bias = (sum_of_y - (m_slope * sum_of_x)) / n
    return bias
```

```python
# slope (m)
def m(n, sum_of_xy, sum_of_x , sum_of_y, sum_of_xs):
    m1 = (n * sum_of_xy)-(sum_of_x * sum_of_y)
    m2 = (n * sum_of_xs) - (sum_of_x) ** 2
    ans = m1/m2
    return ans
```

```python
# sample data
x = np.array([1,2,3,4,5,6,7])
y = np.array([1.5,3.8,6.7,9.0,11.2,13.6,16])
```

```python
sum_of_x = np.sum(x)
sum_of_y = np.sum(y)
n = len(x)
sum_of_xy = np.sum( x * y)
sum_of_xs = np.sum( x ** 2) # sqaure each element and add
sum_of_ys = np.sum( y ** 2)


print(f"Sum of x: {sum_of_x}")
print(f"Sum of y: {sum_of_y}")
print(f"Sum of xy: {sum_of_xy}")
print(f"Sum of x square: {sum_of_xs}")
print(f"Sum of y square: {sum_of_ys}")
print(f"Size of data: {n}")
```

```python
slope_m = m(n, sum_of_xy, sum_of_x , sum_of_y, sum_of_xs)
bias_b = b(sum_of_y, slope_m, sum_of_x,n)
correlation_coefficient(n, sum_of_xy, sum_of_x, sum_of_y, sum_of_xs, sum_of_ys)

print(f"slope m: {slope_m}\nbias b: {bias_b}")
```
