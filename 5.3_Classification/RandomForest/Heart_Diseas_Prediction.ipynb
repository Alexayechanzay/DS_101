## handling null-values

```python
df.columns
```

```python
from sklearn.impute import SimpleImputer
```

### Select only ***str*** values 

```python
df.select_dtypes(include=["int64"]).columns
```

```python
# .tolist() is used to change column obj to list type

categorical_cols = df.select_dtypes(include=["str"]).columns.tolist()
numerical_cols = df.select_dtypes(include = ["float64"]).columns.tolist()
```

```python
categorical_cols
numerical_cols
```

### replacing null-values

```python
# find the median values of each col 
num_imputer = SimpleImputer(strategy="median")
```

```python
# re-fill null-values of each col with respective col's median values
df[numerical_cols] = num_imputer.fit_transform(df[numerical_cols])
```

```python
df.isnull().sum()
```

```python
cat_imputer = SimpleImputer(strategy="most_frequent")
df[categorical_cols] = cat_imputer.fit_transform(df[categorical_cols])
```

```python
df.isnull().sum()
```

## Label Encoding

```python
df.info()
```

```python
categorical_cols
```

## Label encoding using (sklearn LabelEncoder)

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
```

```python
for col in categorical_cols:
    df[col] = le.fit_transform(df[col])
```

```python
df.head()
```

## Model training

```python
df.columns
```

```python
X = df.drop("Heart Disease Status", axis=1)
y = df["Heart Disease Status"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train , X_test, y_train, y_test = train_test_split(X,y,test_size=0.2)
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
model = RandomForestClassifier()
```

```python
model.fit(X_train,y_train)
```

## Model Evaluation

```python
from sklearn.metrics import accuracy_score, classification_report
```

```python
y_cup = model.predict(X_test)
```

```python
print(f"Accuracy score: {accuracy_score(y_test,y_cup)}")
```

```python
print(f"Classification report:\n {classification_report(y_test,y_cup)}")
```
