```python
import matplotlib.pyplot as plt
import random
```

```python
month = [_ for _ in range(1,13,1)]
income_job1 = [ random.randint(1500000,10000000) for _ in range(12)]
income_job2 = [ random.randint(1000000,5000000) for _ in range(12)]

print(month,income_job1,income_job2)
```

## Matplotlib usages (Visualization)

```python
print(plt.style.available) # to show available styles
plt.style.use("fivethirtyeight")

plt.plot(month,income_job1,color="red",linestyle=':',label="Job 1")
plt.plot(month,income_job2,color="g",linestyle='-',label="Job 2")

plt.xlabel("Monthly income in 2026")
plt.ylabel("Income in $")
plt.title("Income from all 2 jobs in 2026")

plt.legend()
plt.grid(1) # 1 or  True 
plt.tight_layout()

plt.show()
```
