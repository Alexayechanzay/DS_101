```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4,]
y = [10,20,30,40]
```

```python
# plt.style.use("")

# .plot() for line graph
# .bar() for bar-chart
# .scatter() for scatter-diagram

plt.scatter(x,y, color="green", marker="X")
plt.title("Line graph")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")

plt.tight_layout()
plt.show()
```
